```python
from pathlib import Path
from PIL import Image # type: ignore

src = Path('C:/Users/mohds/OneDrive/Desktop/MANARASHOPPINGAPP/flutter_app/assets/images/logo.png')
icons_dir = Path('C:/Users/mohds/OneDrive/Desktop/MANARASHOPPINGAPP/flutter_app/web/icons')
icons_dir.mkdir(exist_ok=True)
img = Image.open(src).convert('RGBA')

for name, size in [
    ('Icon-192.png', 192),
    ('Icon-512.png', 512),
    ('Icon.png', 1024),
    ('Icon-maskable-192.png', 192),
    ('Icon-maskable-512.png', 512),
    ('icon-192.png', 192),
]:
    resized = img.resize((size, size), Image.LANCZOS)
    resized.save(icons_dir / name)

bad = icons_dir / 'Icon-maskable-512..png'
if bad.exists():
    bad.unlink()

print('Updated icon files from logo.png')
```
